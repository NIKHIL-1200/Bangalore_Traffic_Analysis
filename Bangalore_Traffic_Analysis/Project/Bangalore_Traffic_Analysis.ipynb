```python
traffic_by_area=(
    df.groupby("Area Name")["Traffic Volume"].sum().sort_values(ascending=False)
)
traffic_by_area
```

```python
plt.figure(figsize=(12,6))
sns.barplot(
    x=top10_traffic.values,
    y=top10_traffic.index
)
plt.title("Top 10 Bangalore Area by Traffic Volume")
plt.xlabel("Traffic Volume")
plt.ylabel("Area")

plt.tight_layout()
plt.show
```

```python
corr_data = df[[
    "Traffic Volume",
    "Average Speed",
    "Travel Time Index",
    "Road Capacity Utilization",
    "Incident Reports",
    "Pedestrian and Cyclist Count"
]]
```

```python
correlation = corr_data.corr()

correlation
```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Traffic Metrics")

plt.tight_layout()
plt.show()
```
